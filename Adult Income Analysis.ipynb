```python
section_heading("Duplicate Records", "♻️")

duplicate_count = df.duplicated().sum()
print("Number of duplicate records:", duplicate_count)
```

```python
df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
```

```python
section_heading("Finding Missing Values", "❓")

missing_summary= pd.DataFrame({"Missing Count": df.isnull().sum()})
print(missing_summary)

df = df.replace("?", np.nan)
df = df.replace(" ?", np.nan)
df = df.replace(" ? ", np.nan)

df.head()
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    df[col] = df[col].str.strip()
```

```python
df = df.replace("?", np.nan)
```

```python
df["income"] = df["income"].str.strip().str.replace(".", "", regex=False)
```

```python
df["income"].value_counts()
```

```python
section_heading("Handling Missing Values", "🧹")

# Separate numerical and categorical columns
numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns
categorical_columns = df.select_dtypes(include=["object"]).columns

# Fill numerical missing values using median
for col in numerical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values using mode
for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after imputation:")
display(df.isnull().sum())
```

```python
section_heading("Checking Data Types", "🔎")

display(df.dtypes)
```

```python
data_type_summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Missing Values": df.isnull().sum().values,
    "Unique Values": df.nunique().values
})

display(data_type_summary)
```

```python
section_heading("Creating Derived Columns", "🧮")

def age_category(age):
    if age < 30:
        return "Young Adult"
    elif age < 50:
        return "Middle Aged"
    else:
        return "Senior Adult"

df["age_group"] = df["age"].apply(age_category)

display(df[["age", "age_group"]].head())
```

```python
def work_category(hours):
    if hours < 35:
        return "Part-time"
    elif hours <= 40:
        return "Full-time"
    else:
        return "Overtime"

df["work_category"] = df["hours-per-week"].apply(work_category)

```

```python
def capital_gain_status(gain):
    if gain > 0:
        return "Has Capital Gain"
    else:
        return "No Capital Gain"

df["capital_gain_status"] = df["capital-gain"].apply(
    capital_gain_status)

display(df[[
    "capital-gain",
    "capital_gain_status"
]].head())
```

```python
def education_category(education_num):
    if education_num <= 8:
        return "Basic Education"
    elif education_num <= 12:
        return "Intermediate Education"
    else:
        return "Advanced Education"

df["education_level"] = df["education-num"].apply(
    education_category
)

display(df[[
    "education-num",
    "education_level"
]].head())
```

```python
def marital_category(status):
    if "Married" in str(status):
        return "Married"
    else:
        return "Not Married"

df["marital_group"] = df["marital-status"].apply(
    marital_category)
```

```python
section_heading("Finding Outliers", "⚠️")

numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

print("Numerical columns:")
print(numeric_columns)
```

```python
outlier_summary = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_mask = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    )

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_mask.sum(),
        "Outlier Percentage": round(outlier_mask.mean() * 100, 2)
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df)
```

```python
limits = {
    "age": (17, 100),
    "education-num": (1, 16),
    "hours-per-week": (1, 98),
    "capital-gain": (0, None),
    "capital-loss": (0, None),
    "fnlwgt": (0, None)
}

for col, (low, high) in limits.items():
    mask = df[col].lt(low)

    if high is not None:
        mask |= df[col].gt(high)

    print(col, ":", mask.sum(), "unrealistic values")
```

```python
section_heading("Handilng Outliers", "⚠️")

invalid_hours = df[(df["hours-per-week"] < 1) | (df["hours-per-week"] > 98)]
print("Number of unrealistic values:", len(invalid_hours))

df.loc[
    (df["hours-per-week"] < 1) | (df["hours-per-week"] > 98),
    "hours-per-week"
] = np.nan
print("Missing values in hours-per-week:", df["hours-per-week"].isnull().sum())

median_hours = df["hours-per-week"].median()
df["hours-per-week"] = df["hours-per-week"].fillna(median_hours)
print("Missing values after handling:", df["hours-per-week"].isnull().sum())

print("Minimum hours:", df["hours-per-week"].min())
print("Maximum hours:", df["hours-per-week"].max())
```

Exploratory Data Analysis (EDA) is performed to understand the structure, distribution, patterns, and relationships present in the dataset. In this project, EDA is used to examine the socio-economic characteristics of adults and identify factors associated with different income levels.

The analysis is divided into three major parts: Univariate analysis, Bivariate analysis, and Multivariate analysis. Univariate analysis examines individual variables, bivariate analysis investigates relationships between two variables, and multivariate analysis examines relationships among multiple variables.
